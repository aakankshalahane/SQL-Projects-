```python
# Import libraries
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

# Create database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create movies table
cursor.execute("""
CREATE TABLE movies (
    movie_id INT,
    title VARCHAR(100),
    genre VARCHAR(50),
    platform VARCHAR(50),
    release_year INT,
    duration_minutes INT,
    rating DECIMAL(3,1),
    reviews INT,
    budget DECIMAL(12,2),
    revenue DECIMAL(12,2),
    views INT,
    language VARCHAR(30)
)
""")

# Insert movie dataset
cursor.execute("""
INSERT INTO movies VALUES
(1,'Inception','Sci-Fi','Netflix',2010,148,8.8,2500000,160000000,839000000,18500000,'English'),
(2,'Interstellar','Sci-Fi','Amazon Prime',2014,169,8.7,2100000,165000000,731000000,16200000,'English'),
(3,'The Dark Knight','Action','Netflix',2008,152,9.0,3200000,185000000,1005000000,22000000,'English'),
(4,'Avengers Endgame','Action','Disney+',2019,181,8.4,2800000,356000000,2798000000,31000000,'English'),
(5,'Parasite','Thriller','Netflix',2019,132,8.5,900000,11400000,258000000,8700000,'Korean'),
(6,'Dangal','Drama','Amazon Prime',2016,161,8.3,700000,7000000,300000000,9200000,'Hindi'),
(7,'3 Idiots','Comedy','Netflix',2009,170,8.4,650000,5500000,90000000,11000000,'Hindi'),
(8,'Joker','Drama','Amazon Prime',2019,122,8.3,1900000,55000000,1074000000,18500000,'English'),
(9,'Avatar','Sci-Fi','Disney+',2009,162,7.8,1400000,237000000,2923000000,28000000,'English'),
(10,'Titanic','Romance','Netflix',1997,195,7.9,1300000,200000000,2260000000,24500000,'English'),
(11,'The Matrix','Sci-Fi','Amazon Prime',1999,136,8.7,1800000,63000000,467000000,14500000,'English'),
(12,'Black Panther','Action','Disney+',2018,134,7.3,850000,200000000,1347000000,17000000,'English'),
(13,'La La Land','Romance','Netflix',2016,128,8.0,650000,30000000,472000000,8200000,'English'),
(14,'RRR','Action','Netflix',2022,187,7.8,420000,7200000,166000000,9800000,'Telugu'),
(15,'Drishyam','Thriller','Amazon Prime',2015,163,8.2,380000,5000000,60000000,7500000,'Hindi'),
(16,'Spider-Man No Way Home','Action','Disney+',2021,148,8.2,950000,200000000,1921000000,26000000,'English'),
(17,'The Social Network','Drama','Netflix',2010,120,7.8,750000,40000000,225000000,6900000,'English'),
(18,'Parasite 2','Thriller','Amazon Prime',2023,140,8.1,320000,18000000,185000000,6100000,'Korean'),
(19,'KGF Chapter 2','Action','Netflix',2022,168,8.3,410000,10000000,130000000,8700000,'Kannada'),
(20,'Coco','Animation','Disney+',2017,105,8.4,850000,175000000,807000000,13500000,'English')
""")

conn.commit()

# View movie dataset
movies = pd.read_sql_query("""
SELECT *
FROM movies
""", conn)

print("Movie Dataset")
display(movies)

# Average rating by genre
genre_ratings = pd.read_sql_query("""
SELECT
    genre,
    COUNT(*) AS movie_count,
    ROUND(AVG(rating), 2) AS average_rating,
    ROUND(AVG(views), 0) AS average_views
FROM movies
GROUP BY genre
ORDER BY average_rating DESC
""", conn)

print("Rating Analysis by Genre")
display(genre_ratings)

# Platform performance
platform_analysis = pd.read_sql_query("""
SELECT
    platform,
    COUNT(*) AS movie_count,
    SUM(views) AS total_views,
    ROUND(AVG(rating), 2) AS average_rating,
    SUM(revenue) AS total_revenue
FROM movies
GROUP BY platform
ORDER BY total_views DESC
""", conn)

print("Platform Performance")
display(platform_analysis)

# Highest rated movies
top_rated = pd.read_sql_query("""
SELECT
    title,
    genre,
    platform,
    rating,
    reviews
FROM movies
WHERE rating >= 8.5
ORDER BY rating DESC, reviews DESC
""", conn)

print("Highest Rated Movies")
display(top_rated)

# Most watched movies
most_watched = pd.read_sql_query("""
SELECT
    title,
    platform,
    genre,
    views
FROM movies
ORDER BY views DESC
LIMIT 10
""", conn)

print("Top 10 Most Watched Movies")
display(most_watched)

# Revenue analysis by genre
genre_revenue = pd.read_sql_query("""
SELECT
    genre,
    COUNT(*) AS movie_count,
    SUM(revenue) AS total_revenue,
    ROUND(AVG(revenue), 2) AS average_revenue
FROM movies
GROUP BY genre
ORDER BY total_revenue DESC
""", conn)

print("Revenue Analysis by Genre")
display(genre_revenue)

# Profit analysis
profit_analysis = pd.read_sql_query("""
SELECT
    title,
    genre,
    budget,
    revenue,
    revenue - budget AS profit
FROM movies
ORDER BY profit DESC
LIMIT 10
""", conn)

print("Top 10 Most Profitable Movies")
display(profit_analysis)

# Movies with high views and high ratings
popular_movies = pd.read_sql_query("""
SELECT
    title,
    genre,
    platform,
    rating,
    views
FROM movies
WHERE rating >= 8
AND views >= 10000000
ORDER BY views DESC
""", conn)

print("Highly Rated and Popular Movies")
display(popular_movies)

# Movies above average revenue
above_average_revenue = pd.read_sql_query("""
SELECT
    title,
    genre,
    revenue
FROM movies
WHERE revenue > (
    SELECT AVG(revenue)
    FROM movies
)
ORDER BY revenue DESC
""", conn)

print("Movies Above Average Revenue")
display(above_average_revenue)

# Language analysis
language_analysis = pd.read_sql_query("""
SELECT
    language,
    COUNT(*) AS movie_count,
    ROUND(AVG(rating), 2) AS average_rating,
    SUM(views) AS total_views,
    SUM(revenue) AS total_revenue
FROM movies
GROUP BY language
ORDER BY total_views DESC
""", conn)

print("Language Analysis")
display(language_analysis)

# Release year analysis
year_analysis = pd.read_sql_query("""
SELECT
    release_year,
    COUNT(*) AS movie_count,
    ROUND(AVG(rating), 2) AS average_rating,
    SUM(views) AS total_views
FROM movies
GROUP BY release_year
ORDER BY release_year
""", conn)

print("Release Year Analysis")
display(year_analysis)

# Longest movies
longest_movies = pd.read_sql_query("""
SELECT
    title,
    genre,
    duration_minutes,
    rating
FROM movies
ORDER BY duration_minutes DESC
LIMIT 5
""", conn)

print("Longest Movies")
display(longest_movies)

# Short movies with high ratings
efficient_movies = pd.read_sql_query("""
SELECT
    title,
    duration_minutes,
    rating,
    genre
FROM movies
WHERE duration_minutes < 140
AND rating >= 8
ORDER BY rating DESC
""", conn)

print("Short Movies With High Ratings")
display(efficient_movies)

# Overall streaming metrics
overall_metrics = pd.read_sql_query("""
SELECT
    COUNT(*) AS total_movies,
    ROUND(AVG(rating), 2) AS average_rating,
    SUM(views) AS total_views,
    SUM(revenue) AS total_revenue,
    SUM(budget) AS total_budget,
    SUM(revenue - budget) AS total_profit
FROM movies
""", conn)

print("Overall Streaming Metrics")
display(overall_metrics)

# Visualize revenue by genre
plt.figure(figsize=(9, 5))
plt.bar(
    genre_revenue["genre"],
    genre_revenue["total_revenue"]
)
plt.title("Total Revenue by Genre")
plt.xlabel("Genre")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.show()

# Visualize platform views
plt.figure(figsize=(9, 5))
plt.bar(
    platform_analysis["platform"],
    platform_analysis["total_views"]
)
plt.title("Total Views by Streaming Platform")
plt.xlabel("Platform")
plt.ylabel("Total Views")
plt.xticks(rotation=30)
plt.show()

# Visualize top movies by views
plt.figure(figsize=(10, 6))
plt.barh(
    most_watched["title"].head(10),
    most_watched["views"].head(10)
)
plt.title("Top 10 Most Watched Movies")
plt.xlabel("Views")
plt.ylabel("Movie")
plt.gca().invert_yaxis()
plt.show()

# Close database
conn.close()

print("Database connection closed successfully")
```
