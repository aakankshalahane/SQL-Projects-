```python
# Customer Churn & Retention Analysis

import sqlite3
import pandas as pd

# Create database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create customers table
cursor.execute("""
CREATE TABLE customers (
    customer_id INT,
    customer_name VARCHAR(50),
    age INT,
    city VARCHAR(50),
    plan VARCHAR(30),
    monthly_fee DECIMAL(10,2),
    months_active INT,
    churned INT
)
""")

# Insert customer data
cursor.executemany("""
INSERT INTO customers VALUES (?, ?, ?, ?, ?, ?, ?, ?)
""", [
    (1, 'Aarav', 24, 'Mumbai', 'Basic', 499, 4, 1),
    (2, 'Priya', 29, 'Delhi', 'Premium', 999, 18, 0),
    (3, 'Rahul', 35, 'Pune', 'Basic', 499, 3, 1),
    (4, 'Sneha', 27, 'Mumbai', 'Premium', 999, 24, 0),
    (5, 'Amit', 41, 'Bangalore', 'Standard', 699, 8, 1),
    (6, 'Neha', 31, 'Delhi', 'Premium', 999, 30, 0),
    (7, 'Rohan', 26, 'Pune', 'Standard', 699, 10, 0),
    (8, 'Isha', 38, 'Mumbai', 'Basic', 499, 2, 1),
    (9, 'Karan', 45, 'Bangalore', 'Premium', 999, 20, 0),
    (10, 'Meera', 28, 'Delhi', 'Standard', 699, 6, 1),
    (11, 'Vikram', 34, 'Mumbai', 'Premium', 999, 16, 0),
    (12, 'Ananya', 23, 'Pune', 'Basic', 499, 5, 1),
    (13, 'Kabir', 39, 'Bangalore', 'Standard', 699, 14, 0),
    (14, 'Pooja', 30, 'Delhi', 'Premium', 999, 22, 0),
    (15, 'Aditya', 25, 'Mumbai', 'Basic', 499, 3, 1)
])

conn.commit()

# Customer data
print("Customer Data")
print("-" * 50)

result = cursor.execute("""
SELECT * FROM customers
""").fetchall()

df = pd.DataFrame(result, columns=[desc[0] for desc in cursor.description])
print(df)

# Total customers
print("\nTotal Customers")
print("-" * 50)

result = cursor.execute("""
SELECT COUNT(*) AS total_customers
FROM customers
""").fetchall()

print(pd.DataFrame(result, columns=["total_customers"]))

# Churned customers
print("\nChurned Customers")
print("-" * 50)

result = cursor.execute("""
SELECT COUNT(*) AS churned_customers
FROM customers
WHERE churned = 1
""").fetchall()

print(pd.DataFrame(result, columns=["churned_customers"]))

# Churn rate
print("\nChurn Rate")
print("-" * 50)

result = cursor.execute("""
SELECT
    ROUND(
        100.0 * SUM(churned) / COUNT(*),
        2
    ) AS churn_rate
FROM customers
""").fetchall()

print(pd.DataFrame(result, columns=["churn_rate_percent"]))

# Churn by plan
print("\nChurn By Plan")
print("-" * 50)

result = cursor.execute("""
SELECT
    plan,
    COUNT(*) AS total_customers,
    SUM(churned) AS churned_customers,
    ROUND(100.0 * SUM(churned) / COUNT(*), 2) AS churn_rate
FROM customers
GROUP BY plan
ORDER BY churn_rate DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "plan",
    "total_customers",
    "churned_customers",
    "churn_rate"
]))

# Churn by city
print("\nChurn By City")
print("-" * 50)

result = cursor.execute("""
SELECT
    city,
    COUNT(*) AS total_customers,
    SUM(churned) AS churned_customers,
    ROUND(100.0 * SUM(churned) / COUNT(*), 2) AS churn_rate
FROM customers
GROUP BY city
ORDER BY churn_rate DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "city",
    "total_customers",
    "churned_customers",
    "churn_rate"
]))

# Average months active
print("\nAverage Months Active")
print("-" * 50)

result = cursor.execute("""
SELECT
    ROUND(AVG(months_active), 2) AS average_months_active
FROM customers
""").fetchall()

print(pd.DataFrame(result, columns=["average_months_active"]))

# Revenue from active customers
print("\nMonthly Revenue From Active Customers")
print("-" * 50)

result = cursor.execute("""
SELECT
    SUM(monthly_fee) AS monthly_revenue
FROM customers
WHERE churned = 0
""").fetchall()

print(pd.DataFrame(result, columns=["monthly_revenue"]))

# Customer lifetime value
print("\nEstimated Customer Lifetime Value")
print("-" * 50)

result = cursor.execute("""
SELECT
    customer_name,
    plan,
    monthly_fee,
    months_active,
    monthly_fee * months_active AS lifetime_value
FROM customers
ORDER BY lifetime_value DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "customer_name",
    "plan",
    "monthly_fee",
    "months_active",
    "lifetime_value"
]))

# High risk customers
print("\nHigh Risk Customers")
print("-" * 50)

result = cursor.execute("""
SELECT
    customer_name,
    plan,
    months_active,
    monthly_fee
FROM customers
WHERE churned = 1
ORDER BY months_active
""").fetchall()

print(pd.DataFrame(result, columns=[
    "customer_name",
    "plan",
    "months_active",
    "monthly_fee"
]))

# Close database
conn.close()
```
