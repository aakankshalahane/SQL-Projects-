```python
# Import libraries
import sqlite3
import pandas as pd

# Create database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create employees table
cursor.execute("""
CREATE TABLE employees (
    employee_id INT,
    employee_name VARCHAR(50),
    department VARCHAR(50),
    job_role VARCHAR(50),
    age INT,
    gender VARCHAR(20),
    salary DECIMAL(10,2),
    experience_years INT,
    performance_score DECIMAL(3,1),
    satisfaction_score DECIMAL(3,1),
    joining_date DATE,
    overtime_hours INT,
    attrition VARCHAR(10)
)
""")

# Insert employee dataset
cursor.execute("""
INSERT INTO employees VALUES
(1,'Aarav','IT','Data Scientist',25,'Male',75000,2,8.5,8.2,'2023-06-10',12,'No'),
(2,'Priya','HR','HR Manager',32,'Female',68000,7,8.1,7.8,'2019-03-15',5,'No'),
(3,'Rahul','IT','Software Engineer',28,'Male',82000,5,9.0,8.7,'2021-01-20',18,'No'),
(4,'Sneha','Finance','Financial Analyst',30,'Female',72000,6,8.4,8.0,'2020-07-12',10,'No'),
(5,'Amit','Sales','Sales Executive',26,'Male',55000,3,7.2,6.5,'2022-05-18',25,'Yes'),
(6,'Neha','IT','Data Analyst',27,'Female',70000,4,8.8,8.5,'2021-09-25',8,'No'),
(7,'Rohan','Marketing','Marketing Manager',35,'Male',90000,10,9.1,8.9,'2017-02-10',6,'No'),
(8,'Isha','Sales','Sales Manager',34,'Female',85000,9,8.6,7.2,'2018-11-05',20,'Yes'),
(9,'Karan','IT','DevOps Engineer',29,'Male',88000,6,9.2,8.8,'2020-04-22',15,'No'),
(10,'Meera','HR','Recruiter',27,'Female',58000,4,7.8,7.5,'2022-08-14',7,'No'),
(11,'Vikram','Finance','Accountant',31,'Male',62000,6,7.9,7.1,'2020-12-01',12,'Yes'),
(12,'Ananya','Marketing','Content Strategist',26,'Female',60000,3,8.3,8.1,'2023-01-16',9,'No'),
(13,'Aditya','IT','Machine Learning Engineer',30,'Male',95000,7,9.5,9.0,'2019-08-20',16,'No'),
(14,'Kavya','Sales','Business Development Executive',25,'Female',52000,2,7.0,6.2,'2023-04-11',28,'Yes'),
(15,'Sahil','Finance','Finance Manager',38,'Male',98000,13,9.0,8.6,'2015-06-30',5,'No'),
(16,'Pooja','HR','HR Executive',29,'Female',61000,5,8.0,7.9,'2021-03-19',4,'No'),
(17,'Manish','IT','Software Engineer',27,'Male',79000,4,8.7,8.3,'2022-02-28',14,'No'),
(18,'Riya','Marketing','SEO Specialist',28,'Female',57000,4,7.6,7.0,'2021-11-09',11,'Yes'),
(19,'Arjun','IT','Data Engineer',31,'Male',92000,8,9.3,8.9,'2018-09-17',17,'No'),
(20,'Simran','Sales','Sales Executive',24,'Female',50000,1,6.9,6.0,'2024-01-10',30,'Yes')
""")

conn.commit()

# View employee data
employees = pd.read_sql_query("""
SELECT *
FROM employees
""", conn)

display(employees)

# Department analysis
department_summary = pd.read_sql_query("""
SELECT
    department,
    COUNT(*) AS employee_count,
    ROUND(AVG(salary), 2) AS average_salary,
    ROUND(AVG(performance_score), 2) AS average_performance,
    ROUND(AVG(satisfaction_score), 2) AS average_satisfaction
FROM employees
GROUP BY department
ORDER BY average_salary DESC
""", conn)

display(department_summary)

# Salary analysis by job role
salary_by_role = pd.read_sql_query("""
SELECT
    job_role,
    COUNT(*) AS employees,
    ROUND(AVG(salary), 2) AS average_salary,
    MAX(salary) AS highest_salary,
    MIN(salary) AS lowest_salary
FROM employees
GROUP BY job_role
ORDER BY average_salary DESC
""", conn)

display(salary_by_role)

# High performing employees
high_performers = pd.read_sql_query("""
SELECT
    employee_name,
    department,
    job_role,
    performance_score,
    salary
FROM employees
WHERE performance_score >= 9
ORDER BY performance_score DESC
""", conn)

display(high_performers)

# Employees earning above average salary
high_salary_employees = pd.read_sql_query("""
SELECT
    employee_name,
    department,
    job_role,
    salary,
    experience_years
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
)
ORDER BY salary DESC
""", conn)

display(high_salary_employees)

# Departments with at least 3 employees
department_hiring = pd.read_sql_query("""
SELECT
    department,
    COUNT(*) AS employee_count
FROM employees
GROUP BY department
HAVING COUNT(*) >= 3
ORDER BY employee_count DESC
""", conn)

display(department_hiring)

# Attrition analysis
attrition_analysis = pd.read_sql_query("""
SELECT
    department,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    ROUND(
        100.0 * SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY department
ORDER BY attrition_rate DESC
""", conn)

display(attrition_analysis)

# Employees with high overtime
overtime_analysis = pd.read_sql_query("""
SELECT
    employee_name,
    department,
    overtime_hours,
    satisfaction_score,
    attrition
FROM employees
WHERE overtime_hours >= 20
ORDER BY overtime_hours DESC
""", conn)

display(overtime_analysis)

# Experience analysis
experience_analysis = pd.read_sql_query("""
SELECT
    CASE
        WHEN experience_years <= 3 THEN '0-3 Years'
        WHEN experience_years <= 7 THEN '4-7 Years'
        ELSE '8+ Years'
    END AS experience_group,
    COUNT(*) AS employee_count,
    ROUND(AVG(salary), 2) AS average_salary,
    ROUND(AVG(performance_score), 2) AS average_performance
FROM employees
GROUP BY experience_group
ORDER BY average_salary DESC
""", conn)

display(experience_analysis)

# Gender analysis
gender_analysis = pd.read_sql_query("""
SELECT
    gender,
    COUNT(*) AS employee_count,
    ROUND(AVG(salary), 2) AS average_salary,
    ROUND(AVG(performance_score), 2) AS average_performance
FROM employees
GROUP BY gender
""", conn)

display(gender_analysis)

# Top 5 employees
top_employees = pd.read_sql_query("""
SELECT
    employee_name,
    department,
    job_role,
    performance_score,
    satisfaction_score,
    salary
FROM employees
ORDER BY performance_score DESC, satisfaction_score DESC
LIMIT 5
""", conn)

display(top_employees)

# Overall HR metrics
overall_metrics = pd.read_sql_query("""
SELECT
    COUNT(*) AS total_employees,
    ROUND(AVG(age), 2) AS average_age,
    ROUND(AVG(salary), 2) AS average_salary,
    ROUND(AVG(experience_years), 2) AS average_experience,
    ROUND(AVG(performance_score), 2) AS average_performance,
    ROUND(AVG(satisfaction_score), 2) AS average_satisfaction,
    SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END) AS total_attrition
FROM employees
""", conn)

display(overall_metrics)

# Import visualization library
import matplotlib.pyplot as plt

# Visualize average salary
plt.figure(figsize=(9, 5))
plt.bar(
    department_summary["department"],
    department_summary["average_salary"]
)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=30)
plt.show()

# Visualize attrition rate
plt.figure(figsize=(9, 5))
plt.bar(
    attrition_analysis["department"],
    attrition_analysis["attrition_rate"]
)
plt.title("Employee Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=30)
plt.show()

# Close database
conn.close()

print("Database connection closed successfully")
```
