```python
# Banking Transaction & Fraud Analysis

import sqlite3
import pandas as pd

# Create database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create transactions table
cursor.execute("""
CREATE TABLE transactions (
    transaction_id INT,
    customer_name VARCHAR(50),
    account_type VARCHAR(30),
    transaction_type VARCHAR(30),
    amount DECIMAL(10,2),
    city VARCHAR(50),
    transaction_date DATE,
    fraud_flag INT
)
""")

# Insert transaction data
cursor.executemany("""
INSERT INTO transactions VALUES (?, ?, ?, ?, ?, ?, ?, ?)
""", [
    (1, 'Aarav', 'Savings', 'Deposit', 50000, 'Mumbai', '2026-01-02', 0),
    (2, 'Priya', 'Current', 'Withdrawal', 12000, 'Delhi', '2026-01-03', 0),
    (3, 'Rahul', 'Savings', 'Transfer', 75000, 'Pune', '2026-01-05', 1),
    (4, 'Sneha', 'Savings', 'Deposit', 30000, 'Mumbai', '2026-01-07', 0),
    (5, 'Amit', 'Current', 'Withdrawal', 85000, 'Bangalore', '2026-01-08', 1),
    (6, 'Neha', 'Savings', 'Transfer', 15000, 'Delhi', '2026-01-10', 0),
    (7, 'Rohan', 'Current', 'Deposit', 90000, 'Pune', '2026-01-12', 0),
    (8, 'Isha', 'Savings', 'Withdrawal', 60000, 'Mumbai', '2026-01-15', 1),
    (9, 'Karan', 'Savings', 'Transfer', 25000, 'Bangalore', '2026-01-17', 0),
    (10, 'Meera', 'Current', 'Withdrawal', 10000, 'Delhi', '2026-01-20', 0),
    (11, 'Vikram', 'Savings', 'Transfer', 95000, 'Mumbai', '2026-01-21', 1),
    (12, 'Ananya', 'Current', 'Deposit', 45000, 'Pune', '2026-01-23', 0),
    (13, 'Kabir', 'Savings', 'Withdrawal', 7000, 'Bangalore', '2026-01-25', 0),
    (14, 'Pooja', 'Current', 'Transfer', 55000, 'Delhi', '2026-01-27', 0),
    (15, 'Aditya', 'Savings', 'Withdrawal', 110000, 'Mumbai', '2026-01-29', 1),
    (16, 'Aarav', 'Savings', 'Transfer', 18000, 'Mumbai', '2026-02-02', 0),
    (17, 'Priya', 'Current', 'Deposit', 65000, 'Delhi', '2026-02-04', 0),
    (18, 'Rahul', 'Savings', 'Withdrawal', 90000, 'Pune', '2026-02-06', 1),
    (19, 'Sneha', 'Savings', 'Transfer', 22000, 'Mumbai', '2026-02-08', 0),
    (20, 'Amit', 'Current', 'Deposit', 35000, 'Bangalore', '2026-02-10', 0)
])

conn.commit()

# Transaction data
print("Transaction Data")
print("-" * 60)

result = cursor.execute("""
SELECT * FROM transactions
""").fetchall()

df = pd.DataFrame(
    result,
    columns=[desc[0] for desc in cursor.description]
)

print(df)

# Total transaction value
print("\nTotal Transaction Value")
print("-" * 60)

result = cursor.execute("""
SELECT
    SUM(amount) AS total_transaction_value
FROM transactions
""").fetchall()

print(pd.DataFrame(result, columns=["total_transaction_value"]))

# Total transactions
print("\nTotal Number Of Transactions")
print("-" * 60)

result = cursor.execute("""
SELECT
    COUNT(*) AS total_transactions
FROM transactions
""").fetchall()

print(pd.DataFrame(result, columns=["total_transactions"]))

# Transactions by type
print("\nTransactions By Type")
print("-" * 60)

result = cursor.execute("""
SELECT
    transaction_type,
    COUNT(*) AS transaction_count,
    SUM(amount) AS total_amount
FROM transactions
GROUP BY transaction_type
ORDER BY total_amount DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "transaction_type",
    "transaction_count",
    "total_amount"
]))

# Transactions by city
print("\nTransactions By City")
print("-" * 60)

result = cursor.execute("""
SELECT
    city,
    COUNT(*) AS transaction_count,
    SUM(amount) AS total_amount
FROM transactions
GROUP BY city
ORDER BY total_amount DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "city",
    "transaction_count",
    "total_amount"
]))

# Fraud transactions
print("\nFraudulent Transactions")
print("-" * 60)

result = cursor.execute("""
SELECT
    transaction_id,
    customer_name,
    transaction_type,
    amount,
    city,
    transaction_date
FROM transactions
WHERE fraud_flag = 1
ORDER BY amount DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "transaction_id",
    "customer_name",
    "transaction_type",
    "amount",
    "city",
    "transaction_date"
]))

# Fraud rate
print("\nFraud Rate")
print("-" * 60)

result = cursor.execute("""
SELECT
    ROUND(
        100.0 * SUM(fraud_flag) / COUNT(*),
        2
    ) AS fraud_rate
FROM transactions
""").fetchall()

print(pd.DataFrame(result, columns=["fraud_rate_percent"]))

# Fraud amount
print("\nTotal Fraudulent Amount")
print("-" * 60)

result = cursor.execute("""
SELECT
    SUM(amount) AS fraudulent_amount
FROM transactions
WHERE fraud_flag = 1
""").fetchall()

print(pd.DataFrame(result, columns=["fraudulent_amount"]))

# Fraud by city
print("\nFraud By City")
print("-" * 60)

result = cursor.execute("""
SELECT
    city,
    COUNT(*) AS fraud_transactions,
    SUM(amount) AS fraud_amount
FROM transactions
WHERE fraud_flag = 1
GROUP BY city
ORDER BY fraud_amount DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "city",
    "fraud_transactions",
    "fraud_amount"
]))

# High value transactions
print("\nHigh Value Transactions")
print("-" * 60)

result = cursor.execute("""
SELECT
    customer_name,
    transaction_type,
    amount,
    city,
    transaction_date,
    fraud_flag
FROM transactions
WHERE amount >= 50000
ORDER BY amount DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "customer_name",
    "transaction_type",
    "amount",
    "city",
    "transaction_date",
    "fraud_flag"
]))

# Customer transaction summary
print("\nCustomer Transaction Summary")
print("-" * 60)

result = cursor.execute("""
SELECT
    customer_name,
    COUNT(*) AS total_transactions,
    SUM(amount) AS total_amount,
    SUM(fraud_flag) AS fraud_transactions
FROM transactions
GROUP BY customer_name
ORDER BY total_amount DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "customer_name",
    "total_transactions",
    "total_amount",
    "fraud_transactions"
]))

# Risky customers
print("\nCustomers With Fraud Activity")
print("-" * 60)

result = cursor.execute("""
SELECT
    customer_name,
    COUNT(*) AS fraud_transactions,
    SUM(amount) AS fraud_amount
FROM transactions
WHERE fraud_flag = 1
GROUP BY customer_name
ORDER BY fraud_amount DESC
""").fetchall()

print(pd.DataFrame(result, columns=[
    "customer_name",
    "fraud_transactions",
    "fraud_amount"
]))

# Close database
conn.close()
```
